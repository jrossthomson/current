```
import py3Dmol
import glob

files = glob.glob('*.cif')
if files:
    view = py3Dmol.view(width=800, height=600)
    with open(files[0], 'r') as f:
        view.addModel(f.read(), 'cif')
    view.setStyle({'cartoon': {'color': 'spectrum'}})
    view.zoomTo()
    view.show()
```
